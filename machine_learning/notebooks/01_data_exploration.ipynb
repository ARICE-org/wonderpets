```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Database connection
DATABASE_URL = "postgresql://postgres:password@localhost:5432/arice_db"
engine = create_engine(DATABASE_URL)
```

```python
# Load soil data
soil_df = pd.read_sql("SELECT * FROM soil_data", engine)
print(f"Soil data shape: {soil_df.shape}")
soil_df.head()
```

```python
# Load weather data
weather_df = pd.read_sql("SELECT * FROM weather_data", engine)
print(f"Weather data shape: {weather_df.shape}")
weather_df.head()
```

```python
# Load rice varieties
varieties_df = pd.read_sql("SELECT * FROM rice_varieties", engine)
print(f"Rice varieties shape: {varieties_df.shape}")
varieties_df.head()
```

```python
# Data summary statistics
print("=== Soil Data Statistics ===")
soil_df.describe()
```

```python
# Check for missing values
print("=== Missing Values ===")
print("Soil:", soil_df.isnull().sum().sum())
print("Weather:", weather_df.isnull().sum().sum())
print("Varieties:", varieties_df.isnull().sum().sum())
```
